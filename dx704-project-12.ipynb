## Part 3: Create Embedding Vectors

Use BERT to create embeddings for each email in the Enron data set.
You will have to decide how to combine the different columns of the original data set to produce one embedding vector.


Hint: BERT can be run without a GPU, but will be much slower.
Using Google Colab with only a CPU, it runs around 1 embedding per second.
Using Google Colab with the T4 GPU option, it runs around 60 embeddings per second.
Caching is also encouraged to avoid unnecessary reruns.

```python
# ======================================
# -- Part 3: Create Embedding Vectors --
# ======================================

enron_spam = enron_spam_data.copy()

# Join 'Subject' and 'Message' features
enron_spam['Subject'] = enron_spam['Subject'].fillna('')
enron_spam['Message'] = enron_spam['Message'].fillna('')
enron_spam['Full_Text'] = enron_spam['Subject'] + ' ' + enron_spam['Message']

# Batch embedding generator function
def generate_embeddings(dataframe, batch_size=64):
    embeddings_list = []
    texts = dataframe['Full_Text'].tolist()

    # Use tqdm to visualize the progress bar
    for i in tqdm(range(0, len(texts), batch_size), desc="Embedding emails"):
        batch_texts = texts[i:i + batch_size]

        # Tokenize batch
        inputs = tokenizer(batch_texts, padding=True, truncation=True, max_length=512, return_tensors="pt").to(device)

        with torch.no_grad():
            outputs = bert_model(**inputs)
            # CLS token embedding is the first token's hidden state
            cls_embeddings = outputs.last_hidden_state[:, 0, :].cpu().numpy()

            # Round to 3 decimal places
            cls_embeddings = np.round(cls_embeddings, 3)

            for emb in cls_embeddings:
                # Convert to 1D list and encode as JSON
                embeddings_list.append(json.dumps(emb.tolist()))

    return embeddings_list

# Generate and assign embeddings
enron_spam['embedding_vector_json'] = generate_embeddings(enron_spam)
```
