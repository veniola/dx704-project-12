## Part 2: Download BERT Model

We will use a pre-trained BERT model to extract embedding vectors as described in lesson 2.1 this week.
Here is sample code to download the model from [Hugging Face](https://huggingface.co/) and extract one vector.

```python
# You may need to install torch and transformers.
# Google Colab will have these installed already.
#
# pip install transformers torch --upgrade

import torch
from transformers import AutoTokenizer, AutoModel
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

```python
MODEL_NAME = "bert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
bert_model = AutoModel.from_pretrained(MODEL_NAME)
bert_model.to(device)
bert_model.eval()
```

```python
@torch.no_grad()
def embed_text(text):
    batch = [text]
    inputs = tokenizer(batch, padding=True, truncation=True, return_tensors="pt").to(device)
    outputs = bert_model(**inputs)
    # CLS token embedding is the first token's hidden state
    cls_emb = outputs.last_hidden_state[:, 0, :]  # shape: [batch_size, 768]
    return cls_emb.cpu()
```

## Part 3: Create Embedding Vectors

Use BERT to create embeddings for each email in the Enron data set.
You will have to decide how to combine the different columns of the original data set to produce one embedding vector.


Hint: BERT can be run without a GPU, but will be much slower.
Using Google Colab with only a CPU, it runs around 1 embedding per second.
Using Google Colab with the T4 GPU option, it runs around 60 embeddings per second.
Caching is also encouraged to avoid unnecessary reruns.

```python
# ======================================
# -- Part 3: Create Embedding Vectors --
# ======================================

# 1. Prepare text column by joining Subject and Message
enron_spam_data['Subject'] = enron_spam_data['Subject'].fillna('')
enron_spam_data['Message'] = enron_spam_data['Message'].fillna('')
enron_spam_data['Full_Text'] = enron_spam_data['Subject'] + " " + enron_spam_data['Message']

# 2. Batch embedding generator function
def generate_embeddings(dataframe, batch_size=64):
    embeddings_list = []
    texts = dataframe['Full_Text'].tolist()

    # Use tqdm to visualize the progress bar
    for i in tqdm(range(0, len(texts), batch_size), desc="Embedding emails"):
        batch_texts = texts[i:i + batch_size]

        # Tokenize batch
        inputs = tokenizer(batch_texts, padding=True, truncation=True, max_length=512, return_tensors="pt").to(device)

        with torch.no_grad():
            outputs = bert_model(**inputs)
            # Extract the CLS token representation (first token) for each sentence in the batch
            cls_embeddings = outputs.last_hidden_state[:, 0, :].cpu().numpy()

            # Round to 3 decimal places to keep file size under 100MB
            cls_embeddings = np.round(cls_embeddings, 3)

            for emb in cls_embeddings:
                # Convert to a flat 1D python list and encode as JSON
                embeddings_list.append(json.dumps(emb.tolist()))

    return embeddings_list

# 3. Generate and assign embeddings
enron_spam_data['embedding_vector_json'] = generate_embeddings(enron_spam_data)

# 4. Save results to embeddings.tsv.gz in the required format
output_df = enron_spam_data[['Message ID', 'embedding_vector_json']]
output_df.to_csv('embeddings.tsv.gz', sep='\t', index=False, compression='gzip')

print("Saved embeddings.tsv.gz successfully!")
```
